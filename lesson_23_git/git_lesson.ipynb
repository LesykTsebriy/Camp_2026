# What is VCS (Version Control System) and why use it at all?

Version control systems - are software tools that help software teams manage changes to source code over time

1. Enhanced Collaboration:
Simultaneous Development:
VCS allows multiple developers to work on the same project concurrently without overwriting each other's work. 

2. Code Quality and Integrity:
Error Reduction:
VCS provides a complete history of changes, making it easier to identify and correct errors by reverting to previous versions. 

3. Efficient Workflow and Productivity:
Faster Development:
VCS automates tasks like merging and conflict resolution, allowing developers to focus on writing code. 

4. Backup and Recovery:
Data Protection:
VCS acts as a backup of the codebase, protecting against data loss due to hardware failures or other issues. 
Easy Rollbacks:
In case of errors or unexpected issues, VCS allows for easy rollback to a previous stable version of the code.

5. Scalability:
Team Growth: VCS supports team growth by providing tools for managing increasingly complex codebases and facilitating collaboration among larger team

# Brief history of VCS 

### Local Version Control System

A **local version control system** maintains a version database on a single developer's computer. It stores the changes made to version-controlled files as patches—that is, it records only the differences between one version and the next. To revert a file to a previous version, the system reconstructs the file by sequentially applying all relevant patches up to the desired version.

**How it works:**
- Every file change is stored as a patch in the local database.
- To see a previous version, the system adds up all the patches to recreate the file at that point in time.

**Limitations:**
- **No collaboration:** Since the database is local, other developers cannot access it, making collaboration nearly impossible.
- **Risk of data loss:** If the local database is lost or corrupted, all version history and patches are lost as well.
- **Difficult to track changes accurately:** Manual management can lead to errors and loss of work.



### Central Version Control System (CVCS)

A **Central Version Control System (CVCS)** is a version control system where the version database is centralized on a single server that is shared among all developers. This central repository stores all the versioned files and their history, allowing multiple developers to collaborate by checking out and committing changes to the central server.

**Key Characteristics:**
- The version history and files are stored in a central server.
- Developers synchronize their work by communicating with this central repository.
- Enables collaboration by providing a single source of truth for the project.

**Risks and Limitations:**
- **Single Point of Failure:** If the central server goes down, becomes corrupted, or is otherwise inaccessible, the entire team loses access to the version history and the ability to collaborate.
- **Downtime Impact:** During server downtime, developers cannot commit changes, update their local copies, or access previous versions.
- **Scalability Issues:** Performance and availability depend heavily on the central server's capacity and network reliability.

**Summary:**  
Centralized VCS improves collaboration over local VCS by sharing a common repository but introduces risks related to server availability and data loss if the central server fails.

### Distributed Version Control System (DVCS) - Git

Git is a leading example of a **Distributed Version Control System (DVCS)**.

Unlike centralized systems, in Git, each developer's machine holds not just the latest version of the project, but the **entire version history**. When you clone a repository (from services like GitHub, GitLab, or Bitbucket), you receive a full backup of the version database—including all branches, tags, and commit history.

**Key Features:**
- **Full Local Copies:** Every clone is a complete backup of the repository, allowing developers to work independently and offline.
- **No Single Point of Failure:** If the central server goes down, the full project history can be restored from any developer’s local copy.
- **Flexible Collaboration:** Developers can create branches, experiment, and merge changes easily, supporting parallel workflows and rapid feedback.
- **Fast Operations:** Most Git operations, such as committing, branching, and viewing history, are performed locally, making them very fast.

**How it works:**
- Developers clone the repository, gaining a complete copy of all project data and history.
- Changes can be made, committed, and reviewed locally.
- Synchronization with the remote server (push/pull) happens only when needed, enabling both independent and collaborative work.

**Resilience Example:**
> If the central server fails or is lost, the entire repository (including all history and branches) can be restored from any developer’s local clone. This makes Git and other DVCS solutions highly robust compared to centralized systems.

## Recording Changes to the Repository

Remember that each file in your working directory can be in one of two states: tracked or untracked. Tracked files are files that were in the last snapshot, as well as any newly staged files; they can be unmodified, modified, or staged. In short, tracked files are files that Git knows about.

Untracked files are everything else — any files in your working directory that were not in your last snapshot and are not in your staging area. When you first clone a repository, all of your files will be tracked and unmodified because Git just checked them out and you haven’t edited anything.

As you edit files, Git sees them as modified, because you’ve changed them since your last commit. As you work, you selectively stage these modified files and then commit all those staged changes, and the cycle repeats.

## git fetch vs git pull  

- **`git fetch`**  
  Just refreshes your local repository with new data (like new branches, tags, and commits), in other words it updates the remote-tracking branches (origin/main...) with the latest commits from the remote but does **not** change your local branches or working files.

- **`git pull`**  
  Adds new commits to your current branch and updates your working directory, so you immediately see and work with the latest changes from the remote.

```python
!git pull #actually performs git fetch followed by a git merge
```

# Cloning the repo

### Authentication Methods: HTTPS vs SSH

### HTTPS

- Authenticate using your Git hosting account’s **username and password**, or more commonly, a **personal access token (PAT)**.
- Simple to set up—no special configuration required.
- Ideal for beginners or when working in restricted network environments.

---

### SSH

- Authenticate using an **SSH key pair** (public and private keys).
- Generate these keys on your computer and add the **public key** to your Git hosting account.
- After setup, you won’t need to enter credentials for each operation.
- Provides enhanced security and convenience for frequent contributors.

---

### When to Use Each Method

- **Use HTTPS** for the simplest, most compatible setup, especially if you’re new to Git or working behind strict firewalls.
- **Use SSH** for a more secure and convenient workflow after initial setup—ideal for regular contributors and teams. SSH requires a one-time key setup, but saves time and enhances security for ongoing work.
