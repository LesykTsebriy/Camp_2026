# The Landscape of NLP Tasks 🗺️

NLP is more than just counting words. It's about teaching computers to perform useful tasks with language. Here are some of the most common ones:

1. Text Classification:
 - Question: What topic does this text belong to?
 - Examples: Is this email spam or not spam? Is this movie review positive or negative? Is this news article about sports, politics, or finance?
2. Named Entity Recognition (NER):
 - Question: What are the key entities (like people, places, organizations) in this text?
 - Examples: Finding all mentions of Apple Inc. (Organization), Tim Cook (Person), and Cupertino (Place) in an article.
3. Part-of-Speech (POS) Tagging:
 - Question: What is the grammatical role of each word?
 - Examples: Identifying run as a verb and dog as a noun. (We did this with NLTK!)
4. Question Answering (Q&A):
 - Question: Given a context document, can you find the answer to my question?
 - Examples: Feeding a Wikipedia article to a model and asking, "What year was the Eiffel Tower completed?"
5. Text Summarization:
 - Question: Can you give me a short, concise summary of this long document?
 - Examples: Generating a 3-sentence summary of a 10-page research paper.
6. Machine Translation:
 - Question: Can you translate this text from one language to another?
 - Examples: Google Translate, translating from English to Ukrainian.
7. Text Generation:
 - Question: Can you create new text that follows a certain style or topic?
 - Examples: Writing a poem, generating marketing copy, or even writing code.

# Spacy

```python
import spacy

# Load the medium-sized English model with word vectors
nlp = spacy.load("en_core_web_md")

# Let's create some doc objects
doc1 = nlp("The king is a powerful ruler.")
doc2 = nlp("The queen is a wise monarch.")
doc3 = nlp("I ate a delicious apple for breakfast.")

# Get the tokens for specific words
king = doc1[1]
queen = doc2[1]
apple = doc3[4]

# Now, let's check their similarity!
# The similarity score is between 0 and 1
print(f"Similarity between 'king' and 'queen': {king.similarity(queen):.2f}")
print(f"Similarity between 'king' and 'apple': {king.similarity(apple):.2f}")

# You can also get the raw vector for a word
print(f"\\nThe vector for 'king' has {king.vector.shape[0]} dimensions.")
```

```python
import spacy

# Load the model again (or use the one from the previous cell)
nlp = spacy.load("en_core_web_md")

text = "Apple Inc. is looking at buying a U.K. startup for $1 billion in London. It was founded by John Appleseed."

# Process the text with the spaCy pipeline
doc = nlp(text)

print("--- Named Entity Recognition (NER) ---")
# Iterate through the detected entities
for ent in doc.ents:
    # Print the entity text and its label (e.g., ORG for organization)
    print(f"Entity: '{ent.text}', Label: '{ent.label_}'")

print("\\n--- Token Details (POS Tagging & Lemmatization) ---")
# Iterate through the tokens in the document
for token in doc:
    # Print the original text, its lemma (base form), and its part-of-speech tag
    print(f"Token: '{token.text}', Lemma: '{token.lemma_}', POS: '{token.pos_}'")
```
